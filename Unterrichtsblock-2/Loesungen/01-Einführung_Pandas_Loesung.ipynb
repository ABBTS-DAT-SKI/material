# Aufgabe 3:
Wie verhält sich die Sample funktion, wenn du sie diese Mehrfach aufrufen? Kannst du die Ergebnisse der Funktionen reproduzierbar machen?

Schau sich dazu die Dokumentation der Funktion an: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sample.html


````python
# Dein  Code:
````

````python
df.sample(5)
````

````python
df.sample(5, random_state = 42)
````

Der Output verändert sich immer. Weil zufällig Zeilen gezogen werden. Möchte man immer das gleiche Set an Zeilen muss man random_state auf eine Zahl setzen.
