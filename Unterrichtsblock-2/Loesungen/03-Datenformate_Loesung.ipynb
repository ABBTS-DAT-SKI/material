# Zusatzaufgabe 3

Öffne die Datei `../data/datenformat_5.csv` in einem Texteditor und schau dir die Daten an. Wie ist das Format aufgebaut? Wo könnte es Probleme geben? 

Lies die Datei anschliessend mit Pandas ein. Entspricht das DataFrame deinen Erwartungen? Stimmen die Datentypen?


```python
# Dein  Code:
```

Die Datei hat in der Spalte study_id ein x und darum ist die Spalte ein Object. Ähnliches gilt für die Spalten patient_id, gender, age_years und age_months.

```python
import pandas as pd

df = pd.read_csv('../data/datenformat_5.csv', delimiter='\t')

print(df)

print(df.dtypes)
```

```python
# replace x in all collumns with NaN or None. remove the x column
df = df.replace('x', pd.NA)
df['study_id'] = pd.to_numeric(df['study_id'], errors='coerce').astype('Int64')
df['patient_id'] = pd.to_numeric(df['patient_id'], errors='coerce').astype('Int64')
df['age_years'] = pd.to_numeric(df['age_years'], errors='coerce').astype('Int64')
df['age_months'] = pd.to_numeric(df['age_months'], errors='coerce').astype('Int64')

df = df.drop(columns=['x'])

print(df)

print(df.dtypes)
```
