Prüfe die Datentypen. Wandle `co2` in eine Ganzzahl und `timestamp` in eine echte Zeitangabe um. Berechne anschliessend den Zeitpunkt eine Stunde später.

- Datentypen anzeigen
- `timestamp` ist zuerst Text
- Ganzzahl
- Zeitangabe plus eine Stunde

```python
# Dein Code hier
```

```python
df.dtypes
```

```python
type(df["timestamp"].iloc[0])
```

```python
df["co2"] = df["co2"].astype(int)
df.dtypes
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
df.dtypes
```

```python
df["timestamp_plus_1h"] = df["timestamp"] + pd.Timedelta(hours=1)
df[["timestamp", "timestamp_plus_1h"]].head()
```
