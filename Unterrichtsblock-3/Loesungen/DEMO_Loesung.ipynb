### Aufgabe 3
Gib den Spalten sprechende Namen und berechne das Leistungsverhältnis von Raum 1 zu Raum 2 als neues Feature. Wie stark ändert sich die Leistung von Raum 1 von einer Messung zur nächsten?

- Namen anpassen
- neue Spalte aus bestehenden Spalten
- Verhältnis als Feature
- Änderung zum vorherigen Messwert

```python
# Dein Code hier
```

```python
demo_umbenannt = sensordaten.rename(columns={"timestamp": "Zeit", "Room1_Temperatur": "Temp_Raum1"})
demo_umbenannt.head()
```

```python
sensordaten["Viel_Leistung"] = sensordaten["Room1_Leistung"] * 2
sensordaten.head()
```

```python
demo_umbenannt["Leistungsverhältnis"] = demo_umbenannt["Room1_Leistung"] / demo_umbenannt["Room2_Leistung"]
demo_umbenannt[["Zeit", "Room1_Leistung", "Room2_Leistung", "Leistungsverhältnis"]].head()
```

```python
# shift(1) verschiebt die Spalte um eine Zeile: so steht der vorherige Messwert daneben
sensordaten["Leistung_Aenderung"] = sensordaten["Room1_Leistung"] - sensordaten["Room1_Leistung"].shift(1)
sensordaten[["timestamp", "Room1_Leistung", "Leistung_Aenderung"]].head()
```
