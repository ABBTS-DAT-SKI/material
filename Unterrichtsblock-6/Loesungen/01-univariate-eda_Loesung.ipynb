```python
# YOUR CODE HERE
import pandas as pd

data = pd.read_csv("../../data/data_resampled.csv")
data
```

```python
data = data[['timestamp', 'room_id', 'temperature']]
data_clean = data[data['temperature'] > 0]
data_clean
```

### Aufgabe 2
Erstelle aus dem bereinigten Datensatz zwei DataFrames für die Räume 413 und 415.


```python
# Dein Code hier
```

```python
df_room_413 = data_clean[data_clean['room_id'] == 413]
df_room_415 = data_clean[data_clean['room_id'] == 415]
```

### Aufgabe 3
Berechne für beide Räume die deskriptiven Statistiken mit `describe()`. Berechne zusätzlich den Median der Temperatur und vergleiche ihn mit dem Mittelwert: Was bedeutet ein deutlicher Unterschied?

```python
# Dein Code hier
```

```python
df_room_413.describe()
```

```python
df_room_415.describe()
```

```python
print("Raum 413: Mittelwert", df_room_413["temperature"].mean(), "| Median", df_room_413["temperature"].median())
print("Raum 415: Mittelwert", df_room_415["temperature"].mean(), "| Median", df_room_415["temperature"].median())
# Liegen Mittelwert und Median weit auseinander, ist die Verteilung schief oder es gibt Ausreisser.
```
