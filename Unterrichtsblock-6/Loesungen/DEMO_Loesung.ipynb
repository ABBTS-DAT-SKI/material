### Aufgabe 2

Verdichte die Zeitreihe von `15min` auf `1h`.

Verwende eine passende Aggregation für `AC_POWER` und prüfe das Ergebnis anschliessend mit einem einfachen Plot (`df.plot()`).

```python
# Dein Code hier
```

```python
# AC_POWER ist eine Leistung: mittlere Leistung pro Stunde -> mean()
# sum() ergäbe das Vierfache, Energie in kWh wäre sum() * 0.25 h
df_1h = df[["AC_POWER"]].resample("1h").mean()
df_1h
```

```python
# 15min und 1h im gleichen Plot: gleicher Verlauf, weniger Punkte
df["AC_POWER"].plot(label="15min", legend=True)
df_1h["AC_POWER"].plot(style="o-", label="1h", legend=True)
```

### Aufgabe 3

Erhöhe die stündliche Reihe wieder auf `30min`.

Fülle die neu entstehenden Zwischenwerte auf zwei Arten und vergleiche die Resultate in einem einfachen Plot.

- zuerst nur das neue Zeitraster ansehen
- `resample` erzeugt Zeilen, gefüllt wird erst danach

```python
# Dein Code hier
```

```python
# resample ändert nur das Zeitraster: die neuen Zeilen sind NaN
df_1h.resample("30min").asfreq().head(8)
```

```python
df_30min = df_1h.resample("30min").interpolate(method="linear")
df_30min.head(8)
```

```python
# zweite Methode: letzten bekannten Wert halten
df_30min_ffill = df_1h.resample("30min").ffill()

df_30min["AC_POWER"].plot(style="x-", label="linear", legend=True)
df_30min_ffill["AC_POWER"].plot(style=".-", label="ffill", legend=True)
```
