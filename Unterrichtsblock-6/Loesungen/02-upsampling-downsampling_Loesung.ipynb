```python
# YOUR CODE HERE
import pandas as pd

data = pd.read_csv("../../data/co2.csv")
data
```

```python
data["timestamp"] = pd.to_datetime(data["timestamp"])
data = data.set_index("timestamp")
data
```

```python
data.plot()
```

### Aufgabe 2
Führe ein Downsampling auf 4-Stunden-Intervalle durch und prüfe das Ergebnis.


```python
# Dein Code hier
```

```python
data_4h = data.resample("4h").mean()
data_4h
```

```python
data_4h.plot()
```

### Aufgabe 3
Führe ein Upsampling auf 30-Minuten-Intervalle durch und vergleiche mindestens zwei Interpolationsmethoden.


```python
# Dein Code hier
```

```python
data_30min = data_4h.resample("30min").interpolate(method="linear")
data_30min.plot()
```

```python
# zweite Methode: letzten bekannten Wert halten
data_30min_ffill = data_4h.resample("30min").ffill()
data_30min_ffill.plot()
```
