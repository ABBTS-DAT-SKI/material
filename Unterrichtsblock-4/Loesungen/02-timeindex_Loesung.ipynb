#### 6. Winterzeitfalle: doppelte Stunde im Herbst

Beim Wechsel von Sommerzeit auf Winterzeit kommt die Stunde zwischen `02:00` und `02:59` zweimal vor. Deshalb ist diese lokale Uhrzeit ohne Zusatzinformation mehrdeutig. Mit `ambiguous="infer"` nutzt pandas die Reihenfolge der Daten und unterscheidet die beiden Vorkommen.

```python
df_winterzeit = df_winterzeit.tz_localize("Europe/Zurich", ambiguous="infer")
df_winterzeit
```

```python
df_winterzeit = df_winterzeit.tz_convert("UTC")
df_winterzeit
```
