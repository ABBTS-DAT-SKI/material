#### 1. Fehlende Werte (NA-Werte) ersetzen  

```python
# welche spalten sind überhaupt NAs?
df.isna().sum()
```

```python
# Wie sieht die Verteilung innerhalb der Spalten aus? 
df[["DC_POWER", "AC_POWER", "DAILY_YIELD"]].describe()
```

```python
# Es handelt sich ja um Solardaten, und auffällig ist hier, dass mindestens zu Beginn des Tages und am Ende des Tages DC Power und AC Power fehlende Werte haben sowie Daily Yield am ersten Tag. 
# Da es sehr viele Duplikate gibt, folgt zunächst nur eine Stichprobe. Die Duplikate werden später entfernt.

df.sample(10)
```

Es scheint so, als sei Daily Yield zu Beginn jeweils ein fehlender Wert, aber summiert sich dann kumulativ über den Tag. DC und AC Power sind jeweils fehlende Werte, wenn gar keine Sonne scheint und somit auch kein Solarstrom produziert wird. Wir können deshalb diese Werte problemlos durch Null ersetzen. 

```python
df = df.fillna(0)

df.sample(10)
```

Das sieht nun sehr gut aus.

```python
# Wenn wir uns die ersten 25 Zeilen ansehen, dann fällt uns auf, dass mehr als 20 davon Duplikate sind. 
df.head(25)
```

```python
# Wir testen, wie viele Duplikate im Datensatz vorhanden sind. 
duplicate_count = df.duplicated(subset=['DATE_TIME', 'PLANT_ID']).sum()
```

Ein grösserer Teil der Daten ist also dupliziert. Es soll jedoch nur die erste Zeile je Duplikat behalten werden. 

```python
df_deduplicated = df.drop_duplicates(subset=['DATE_TIME', 'PLANT_ID'])
```

```python
df_deduplicated.head(25)
```

Nun sehen wir gut, dass sich die Zeit bereits auf der nächsten Zeile um 15 Minuten erhöht, und wir keine Duplikate mehr haben.
