```python
import pandas as pd
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("../../data/data_resampled.csv")
data["timestamp"] = pd.to_datetime(data["timestamp"])
data = data[data["room_id"] == 415].copy()
data = data[data["light"].notna() & data["co2"].notna()]
data = data[(data["light"] != -1) & (data["co2"] != -1)].copy()
data = data.sort_values("timestamp")

data[["timestamp", "light", "co2"]].head()
```

```python
fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=data["light"],
        y=data["co2"],
        mode="markers",
        name="Messpunkte",
        marker={"opacity": 0.45},
    )
)

fig.update_layout(
    title="Licht und CO2 in Raum 415",
    xaxis_title="Licht",
    yaxis_title="CO2 [ppm]",
)

fig.show()
```

### Aufgabe 1
Schätze die CO2-Konzentration in Raum 415 grob mit dem Lichtsensor.

- light gegen co2
- Gerade als einfache Schätzung
- Steigung und Achsenabschnitt
- Vorhersage als Spalte `prediction`, der Plot ist vorbereitet

```python
data[["light", "co2"]].head()
```

```python
# Dein Code hier
```

```python
model = LinearRegression()
X = data[["light"]]
y = data["co2"]
model.fit(X, y)

data["prediction"] = model.predict(X)
```

```python
print(f"Achsenabschnitt b: {model.intercept_}")
print(f"Steigung m: {model.coef_}")
```

```python
# Plot ist vorbereitet: braucht die Spalte data["prediction"] aus deinem Modell
plot_data = data.sort_values("light")

fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=data["light"],
        y=data["co2"],
        mode="markers",
        name="Messpunkte",
        marker={"opacity": 0.45},
    )
)
fig.add_trace(
    go.Scatter(
        x=plot_data["light"],
        y=plot_data["prediction"],
        mode="lines",
        name="Regressionsgerade",
        line={"color": "crimson", "width": 3},
    )
)

fig.update_layout(
    title="Licht und CO2 mit Regressionsgerade",
    xaxis_title="Licht",
    yaxis_title="CO2 [ppm]",
)

fig.show()
```

### Aufgabe 2
Prüfe, wo das Modell gut passt und wo nicht.

- prediction und echte Werte
- Residuen berechnen
- Punkte um 0

```python
# Residuen sind die Differenz zwischen tatsächlichen und vorhergesagten Werten
# Dein Code hier
```

```python
data["residuals"] = data["co2"] - data["prediction"]
```

```python
# Mittelwert der Residuen liegt bei 0
data["residuals"].mean()
```

```python
fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=data["light"],
        y=data["residuals"],
        mode="markers",
        name="Residuen",
        marker={"opacity": 0.45},
    )
)
fig.add_hline(y=0, line_color="crimson", line_dash="dash")

fig.update_layout(
    title="Residuen des Modells",
    xaxis_title="Licht",
    yaxis_title="Residuen",
)

fig.show()
```

### Aufgabe 3
Fasse den Modellfehler mit zwei Kennzahlen zusammen.

- MAE
- MSE
- Unterschiede kurz deuten

```python
# Dein Code hier
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mse = mean_squared_error(data["co2"], data["prediction"])
mae = mean_absolute_error(data["co2"], data["prediction"])

print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
```
