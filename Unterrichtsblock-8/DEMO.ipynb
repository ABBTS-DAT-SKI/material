```python
import pandas as pd
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("../data/data_resampled.csv")
data["timestamp"] = pd.to_datetime(data["timestamp"])
data = data[data["room_id"] == 415].copy()
data = data[data["light"].notna() & data["co2"].notna()]
data = data[(data["light"] != -1) & (data["co2"] != -1)].copy()
data = data.sort_values("timestamp")

data[["timestamp", "light", "co2"]].head()
```

```python
fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=data["light"],
        y=data["co2"],
        mode="markers",
        name="Messpunkte",
        marker={"opacity": 0.45},
    )
)

fig.update_layout(
    title="Licht und CO2 in Raum 415",
    xaxis_title="Licht",
    yaxis_title="CO2 [ppm]",
)

fig.show()
```
