#### 1. Zeitreihen visualisieren

```python
import pandas as pd

df = pd.read_csv("../../data/data_resampled.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df[df["room_id"] == 415]
df.head(3)
```

```python
import plotly.graph_objs as go

fig = go.Figure()
fig.add_trace(go.Scatter(x=df["timestamp"], y=df['co2'], mode='lines'))

fig.update_layout(
    title='CO2 im Raum 415',
    xaxis_title='Datum',
    yaxis_title='CO2',
)

fig.show()
```

**Welche Muster/Auffälligkeiten findest du?**
- Es gibt 5 Peaks und an zwei Stellen fehlen Werte.
- Die Peaks sind meistens am Abend (20:00-23:00).

#### 2. Verteilung visualisieren

```python
fig_hist = go.Figure()
fig_hist.add_trace(go.Histogram(x=df['co2'], nbinsx=20)) # nbinsx=5 wäre zu wenig

fig_hist.update_layout(
    title='Histogram von CO2 Messwerten',
    xaxis_title='CO2',
    yaxis_title='Count'
)
fig_hist.show()
```

**Welche Muster/Auffälligkeiten findest du?**
- Viele CO2-Werte bewegen sich im Bereich von 450 und nach oben. Höhere CO2-Werte gibt es aber abnehmend weniger.
- Es ist gut zu erkennen, dass es viele Werte um 0 herum gibt, aber keine Werte zwischen 450 und 0.
