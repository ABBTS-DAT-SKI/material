### Aufgabe 1
Der durchschnittliche CO2-Messwert pro Raum ist in `avg_co2_by_room` bereits vorbereitet. Erstelle daraus einen Barplot mit beiden Räumen.

```python
df = pd.read_csv("../../data/data_resampled.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["room_id"] = df["room_id"].astype("string")  # damit die x-Achse nicht als Zahl interpretiert wird

# Durchschnittlicher CO2-Wert pro Raum
avg_co2_by_room = df.groupby("room_id")["co2"].mean().reset_index()
avg_co2_by_room
```

```python
# Dein Code hier
```

#### 1. Zwei Variablen visualisieren (numerische und kategorische)

```python
import plotly.graph_objs as go

fig = go.Figure()
fig.add_trace(go.Bar(
    x=avg_co2_by_room["room_id"],
    y=avg_co2_by_room["co2"],
))

fig.update_layout(
    title='Durchschnittlicher CO2-Wert pro Raum',
    xaxis_title='Raum ID',
    yaxis_title='Durchschnittlicher CO2-Wert',
)

fig.show()
```

#### 2. Temperatur und Luftfeuchtigkeit praktisch lesen


```python
import plotly.graph_objs as go

df = pd.read_csv("../../data/room_413.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df["temperature"],
    y=df["humidity"],
    mode="markers"
))

fig.update_layout(
    title="Punktewolke: Temperatur vs. Luftfeuchtigkeit",
    xaxis_title="Temperatur",
    yaxis_title="Luftfeuchtigkeit",
)

fig.show()
```

```python
mean_temperature = df["temperature"].mean()
{
    "humidity_kuehler": round(df.loc[df["temperature"] < mean_temperature, "humidity"].mean(), 2),
    "humidity_waermer": round(df.loc[df["temperature"] >= mean_temperature, "humidity"].mean(), 2),
}
```

Welche betriebliche Folgerung ist sinnvoll?
- Mit höherer Temperatur sinkt die relative Luftfeuchtigkeit.
- In wärmeren Phasen ist die Luft im Mittel trockener.
- Praktische Folgerung: Bei warmen Phasen Feuchtigkeitszufuhr oder Befeuchtung früh prüfen.

#### 4. Pearson und Spearman vergleichen


```python
import plotly.graph_objs as go

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df["temperature"],
    y=df["co2"],
    mode="markers"
))

fig.update_layout(
    title="Punktewolke: Temperatur vs. CO2",
    xaxis_title="Temperatur",
    yaxis_title="CO2",
)

fig.show()
```

```python
df[["temperature", "co2"]].corr(method="pearson")
```

```python
df[["temperature", "co2"]].corr(method="spearman")
```

- Spearman passt besser, wenn die Punktewolke eher monoton als sauber linear wirkt.
