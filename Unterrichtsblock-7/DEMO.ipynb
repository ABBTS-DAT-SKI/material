## Teil 2: Bivariate Visualisierungen

### Aufgabe 3
Vergleiche die durchschnittliche CO2-Belastung zwischen `Nacht`, `Morgen`, `Nachmittag` und `Abend` mit einem Barplot.

Beantworte anschliessend kurz:

- In welcher Tageszeit ist die mittlere CO2-Belastung am höchsten?
- Passt das zu deiner Erwartung für einen genutzten Raum?

```python
def tageszeit(stunde):
    if stunde <= 5:
        return "Nacht"
    elif stunde <= 11:
        return "Morgen"
    elif stunde <= 17:
        return "Nachmittag"
    else:
        return "Abend"


df_co2["tageszeit"] = df_co2["timestamp"].dt.hour.apply(tageszeit)

avg_co2_by_time = (
    df_co2.groupby("tageszeit", observed=False)["co2"].mean().reset_index()
)

avg_co2_by_time
```
